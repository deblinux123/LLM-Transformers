```python
from datasets import load_dataset

data_files = {"train": "drugsComTrain_raw.tsv", "test":"drugsComTest_raw.tsv"}

drug_dataset = load_dataset("csv", data_files=data_files, delimiter="\t")

drug_dataset
```

```python
drug_sample = drug_dataset["train"].shuffle(seed=42).select(range(1000))

drug_sample[:3]
```

```python
for split in drug_dataset.keys():
    assert len(drug_dataset[split]) == len(drug_dataset[split].unique("Unnamed: 0"))
```

```python
drug_dataset = drug_dataset.rename_column(
    original_column_name="Unnamed: 0",
    new_column_name="patient_id"
)

drug_dataset
```

```python
drug_dataset["train"]["patient_id"]
```

```python
drug_dataset["train"]["condition"]
```

```python
def lowercase_condition(example):
    return {"condition":example["condition"].lower()}

drug_dataset.map(lowercase_condition)
```

```python
def filter_nones(x):
    return x["condition"] is not None
```

## Creating new columns

```python
def compute_review_length(example):
    return {"review_length": len(example["review"].split())}
```

```python
drug_dataset = drug_dataset.map(compute_review_length)

drug_dataset["train"][0]
```

```python
drug_dataset["train"].sort("review_length")[:3]
```

```python
drug_dataset = drug_dataset.filter(lambda x: x["review_length"] > 30)
drug_dataset.num_rows
```

```python
import html

text = "I&#039;m a transformer called BERT"
html.unescape(text)
```

```python
drug_dataset = drug_dataset.map(lambda x : {"review":html.unescape(x["review"])})
drug_dataset
```

```python
drug_dataset["train"]["review"][:2]
```

```python
new_drug_dataset = drug_dataset.map(
    lambda x : {"review":[html.unescape(o) for o in x["review"]]}, batched=True
)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

def tokenize_function(example):
    return tokenizer(example["review"], truncation=True)
```

```python
%time tokenized_dataset = drug_dataset.map(tokenize_function, batched=True)
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased", use_fast=False)

def tokenize_function(example):
    return tokenizer(example["review"], truncation=True)
```

```python
%time tokenized_dataset = drug_dataset.map(tokenize_function)
```

```python
%time
slow_tokenizer = AutoTokenizer.from_pretrained("bert-base-cased", use_fast=False)

def slow_tokenize_function(examples):
    return slow_tokenizer(examples["review"], truncation=True)


tokenized_dataset = drug_dataset.map(slow_tokenize_function, batched=True, num_proc=9)
```

```python
def tokenize_and_split(examples):
    return tokenizer(
        examples["review"],
        truncation=True,
        max_length=128,
        return_overflowing_tokens=True,
    )
```

```python
result = tokenize_and_split(drug_dataset["train"][0])
s = [inp for inp in result["input_ids"]]

len(s)
```

```python
tokenized_dataset = drug_dataset.map(tokenize_and_split, batched=True)
```

## From Datasets to DataFrames and back

```python
drug_dataset.set_format("pandas")
drug_dataset["train"][:3]
```

```python
train_df = drug_dataset["train"][:]
```

```python
train_df
```

```python
frequencies = (
    train_df["condition"].value_counts().to_frame().reset_index().rename(columns={"index":"condition", "count": "frequencies"})
)
```

```python
frequencies.head()
```

```python
from datasets import Dataset

freq_dataset = Dataset.from_pandas(frequencies)
freq_dataset
```

## Create a Validation set

```python
drug_dataset_clean = drug_dataset["train"].train_test_split(train_size=0.8, seed=42)

drug_dataset_clean["validation"] = drug_dataset_clean.pop("test")

drug_dataset_clean["test"] = drug_dataset["test"]

drug_dataset_clean
```
