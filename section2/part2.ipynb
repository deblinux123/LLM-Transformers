# Putting it all togather

```python
from transformers import AutoTokenizer

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)

sequence = "Hello, this is the test for tokenizer."

model_inputs = tokenizer(sequence)
```

```python
sequences = ["I've been waiting for a HuggingFace course my whole life.", "So have I!"]

model_inputs = tokenizer(sequences)
```

```python
# will pad the sequences up to the maximum sequence length
model_inputs = tokenizer(sequence, padding="longest")

# will pad the sequences up to the model max length
model_inputs = tokenizer(sequence, padding="max_length")

# will padd the sequences up to the specified max length
model_inputs = tokenizer(sequence, padding="max_length", max_length=8)
```

```python
sequences = ["I've been waiting for a HuggingFace course my whole life.", "So have I!"]

model_inputs = tokenizer(sequence, truncation=True)

model_inputs = tokenizer(sequence, max_length=8, truncation=True)
```

## Special tokens

```python
sequence = "Hello my name is farid and i love to work with huggingface transformers."

model_inputs = tokenizer(sequence)

print(model_inputs['input_ids'])

tokens = tokenizer.tokenize(sequence)
ids = tokenizer.convert_tokens_to_ids(tokens)
print(ids)
```

```python
print(tokenizer.decode(model_inputs['input_ids']))
print(tokenizer.decode(ids))
```

## Wrap up: from tokenizer to model

```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification

tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = AutoModelForSequenceClassification.from_pretrained(checkpoint)

sequence = ["I love to work with huggingface transformers", "So have I!"]

tokens = tokenizer(sequence,
                   padding=True,
                   trucation=True,
                   return_tensors="pt")
output = model(**tokens)
```
